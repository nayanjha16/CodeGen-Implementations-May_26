# Dataset Exploration

Explore NL2Py and Java2Py preprocessed datasets.

```python
from pathlib import Path
from datasets import load_from_disk

PROJECT_ROOT = Path("..").resolve()

nl2py_train = load_from_disk(str(PROJECT_ROOT / "data/processed/nl2py/train"))
java2py_train = load_from_disk(str(PROJECT_ROOT / "data/processed/java2py/train"))

print(f"NL2Py train samples: {len(nl2py_train)}")
print(f"Java2Py train samples: {len(java2py_train)}")
```

```python
print("=== NL2Py sample ===")
print(nl2py_train[0]["text"][:500])
print("\n=== Java2Py sample ===")
print(java2py_train[0]["text"][:500])
```

```python
from collections import Counter

nl_sources = Counter(nl2py_train["source"])
java_sources = Counter(java2py_train["source"])
print("NL2Py sources:", dict(nl_sources))
print("Java2Py sources:", dict(java_sources))
```
